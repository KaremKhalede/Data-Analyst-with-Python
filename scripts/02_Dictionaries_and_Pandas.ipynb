## Lesson 2.1: Introduction to Dictionaries

In this section, I transition from standard Python **Lists** to **Dictionaries** (`dict`). 

### Key Discoveries:
* **The Problem with Lists**: Accessing elements via numerical indices (e.g., `list.index()`) becomes inefficient and error-prone as datasets grow.
* **The Dictionary Solution**: Storing data using a **Key-Value** pair system inside curly braces `{}` allows for direct, labeled data retrieval.
* **Functions Practiced**: 
  * `dict.keys()`: To fetch all custom labels (Keys).
  * `dict['key_name']`: To immediately retrieve the underlying data (Value) without iteration.


```python
# --- Method 1: The Complex Way (Using Lists) ---
countries =['spain', 'france', 'germany', 'norway']
capitals =['madrid', 'paris', 'berlin', 'oslo']

# Find the index of Germany and print its capital
ind_ger = countries.index('germany')
print("Using Lists (Germany's Capital):", capitals[ind_ger])
```

```python
# --- Method 2: The Professional Way (Using Dictionaries) ---
# Defining dictionary where Country is Key, Capital is Value
europe = {'spain':'madrid', 'france':'paris', 'germany':'berlin', 'norway':'oslo'}

# Print all available keys
print("\nAvailable Keys:", europe.keys())

# Fetch Germany's capital directly using its key label
print("Using Dictionary (Germany's Capital):", europe['germany'])
```

## Lesson 2.2: Advanced Dictionary Manipulations & Nested Dictionaries

In this section, I practice advanced operations on Python dictionaries:
* **Adding & Updating Data**: Assigning a new value directly to a key label (e.g., `dict['new_key'] = value`).
* **Nested Dictionaries**: Creating a "Dictionary of Dictionaries" to add multiple data layers (like capital and population) under a single unique key.


```python
# --- 1. Nested Dictionaries (Dictionary of Dictionaries) ---
europe_nested = { 'spain': { 'capital':'madrid', 'population':46.77 },
                  'france': { 'capital':'paris', 'population':66.03 },
                  'germany': { 'capital':'berlin', 'population':80.62 },
                  'norway': { 'capital':'oslo', 'population':5.084 } }

# Accessing nested data
print("\nFrance's Capital (Nested):", europe_nested['france']['capital'])

```

```python
# --- 2. Add new entry (Italy) ---
italy_data = {'capital': 'rome', 'population': 59.83}
europe_nested['italy'] = italy_data

print("\nUpdated European Database (Nested):\n", europe_nested)
```

```python
# --- 3. Delete entry (Italy) ---
# Check if italy exists before deleting to avoid errors
if 'italy' in europe_nested:
    del(europe_nested['italy'])

print("\nUpdated European Database (Nested):\n", europe_nested)
```

## Lesson 2.3: Introduction to Pandas & DataFrames

In this section, I transition to **Pandas**, the core library for data manipulation in Python, to work with structured tabular data called **DataFrames**.

### Key Methods Practiced:
* **DataFrame from Dictionary**: Converting a standard Python dictionary of lists into a table using `pd.DataFrame()`.
* **Custom Row Indexing**: Assigning distinct row labels by updating the `.index` property.
* **Importing CSV Files**: Utilizing `pd.read_csv()` to import real-world datasets and configuring `index_col=0` to correctly transform the first column into the table's index.


```python
import pandas as pd

# --- Method 1: Creating a DataFrame from a Dictionary ---
names = ['United States', 'Australia', 'Japan', 'India', 'Russia', 'Morocco', 'Egypt']
dr = [True, False, False, False, True, True, True]
cpc = [809, 731, 588, 18, 200, 70, 45]

# Build dictionary and convert to DataFrame
cars_dict = { 'country': names, 'drives_right': dr, 'cars_per_cap': cpc }
cars_manual = pd.DataFrame(cars_dict)

# Fix row labels using custom index names
cars_manual.index = ['US', 'AUS', 'JPN', 'IN', 'RU', 'MOR', 'EG']
print("DataFrame created manually from Dictionary:\n", cars_manual)
```

## Lesson 2.4: Indexing and Selecting Data from DataFrames

In this section, I practice various techniques to subset, slice, and extract specific observations or variables from a Pandas DataFrame using the manually created `cars_manual` dataset.

### Key Indexing Methods Covered:
1. **Square Brackets `[]`**: 
   * Single bracket `['col']` to extract a column as a **Pandas Series**.
   * Double brackets `[['col']]` to extract it as a **Pandas DataFrame**.
   * Index slicing `[start:end]` to subset rows (observations).
2. **`loc` (Label-Based Indexing)**: Selecting rows and columns using their specific textual names/labels.
3. **`iloc` (Integer-Based Indexing)**: Selecting rows and columns using their numerical position/index (0-indexed).


```python
# --- 1. Basic Indexing with Square Brackets [] ---
# Extracting columns as Series vs DataFrame
country_series = cars_manual['country']
country_df = cars_manual[['country']]

print("1. Column as Series (Single Bracket):\n", country_series.head(2))
print("\n2. Column as DataFrame (Double Brackets):\n", country_df.head(2))

# Row slicing (Extracting 4th, 5th, and 6th rows)
print("\n3. Slicing Rows [3:6]:\n", cars_manual[3:6])
```

```python
# --- 2. Advanced Indexing with loc and iloc ---
# Using loc (Label-based) to find Morocco's drive orientation
morocco_drive = cars_manual.loc['MOR', 'drives_right']
print("\n4. Label Selection (Morocco drives_right):", morocco_drive)

# Using iloc (Integer-based) to fetch a sub-DataFrame for Australia and Egypt
# Australia is at index position 1, Egypt is at index position 6
sub_df = cars_manual.iloc[[1, 6], [0, 1]] 
print("\n5. Position Selection (Sub-DataFrame using iloc):\n", sub_df)
```
