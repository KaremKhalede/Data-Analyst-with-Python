## Lesson 2: Boolean Operators

In this lesson, I learned how to combine multiple conditions using Python's built-in boolean operators (`and`, `or`, `not`). Additionally, I explored how to perform element-wise logical operations on NumPy arrays using dedicated functions.

### Objectives:
* Use `and`, `or`, and `not` to evaluate multiple scalar conditions.
* Understand the difference between scalar logic and array logic.
* Utilize NumPy functions (`np.logical_and()`, `np.logical_or()`) for element-wise array filtering.


```python
# 1. Scalar logic with multiple conditions
my_kitchen = 18.0
your_kitchen = 14.0

print(my_kitchen > 10 and my_kitchen < 18)
print(my_kitchen < 14 or my_kitchen > 17)

```

```python
# 2. Array logic using NumPy functions
my_house = np.array([18.0, 20.0, 10.75, 9.50])
your_house = np.array([14.0, 24.0, 14.25, 9.0])

print(np.logical_or(my_house > 18.5, my_house < 10))
print(np.logical_and(my_house < 11, your_house < 11))
```

## Lesson 3: Conditional Statements (Control Flow)

In this lesson, I learned how to use conditional statements (`if`, `elif`, `else`) to guide the execution path of a program based on different conditions.

### Objectives:
* Use `if` to execute code only when a condition is true.
* Use `else` to provide an alternative execution path when the condition fails.
* Use `elif` to check multiple successive conditions sequentially.


```python
# Define variables
room = "kit"
area = 14.0

# 1. First condition construct for room
if room == "kit":
    print("looking around in the kitchen.")
else:
    print("looking around elsewhere.")
```

```python
# 2. Second condition construct for area
if area > 15:
    print("big place!")
elif area > 10:
    print("medium place!")
else:
    print("pretty small.")
```

## Lesson 4: Filtering DataFrames

In this lesson, I learned how to filter and subset DataFrames in Pandas based on logical conditions. Additionally, I explored how to combine multiple criteria using NumPy's logical operators (`np.logical_and`, `np.logical_or`, `np.logical_not`) for advanced filtering.

### Objectives:
* Extract columns as Pandas Series to create boolean vectors.
* Use boolean series to subset and filter specific observations from a DataFrame.
* Combine multiple logical conditions using NumPy functions for element-wise array logic.


```python
# Load the dataset
cars = pd.read_csv('../datasets/cars.csv', index_col = 0)

# 1. Basic filtering (Example: countries that drive on the right)
dr = cars['drives_right']
sel = cars[dr]
print("--- Driving Right Observations ---")
print(sel)
print("\n")
```

```python
# 2. Advanced filtering with NumPy (Example: cars_per_cap between 100 and 500)
cpc = cars['cars_per_cap']
medium = cars[np.logical_and(cpc >= 100, cpc <= 500)]
print("--- Medium Cars Per Cap Observations ---")
print(medium)
```
