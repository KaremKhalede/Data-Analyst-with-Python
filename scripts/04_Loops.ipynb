## 1. The `while` Loop

A `while` loop repeatedly checks a condition before executing the code block. To prevent an **infinite loop**, the variables within the condition must be updated during each iteration.

### Exercise 1.1: Basic while Loop
* **Objective:** Create a loop that decrements a positive `offset` variable until it reaches `0`.


```python
# Initialize offset
offset = 8

# Code the while loop
while offset != 0:
    print("correcting...")
    offset = offset - 1
    print(offset)
```

### Exercise 1.2: Adding Conditional Logic inside a while Loop
* **Objective:** Enhance the loop to handle both positive and negative values of `offset`, safely moving the value towards `0` regardless of its starting sign.

```python
# Initialize negative offset
offset = -6

# Code the while loop with if-else logic
while offset != 0:
    print("correcting...")
    if offset > 0:
        offset = offset - 1
    else:
        offset = offset + 1
    print(offset)
```

## 2. The `for` Loop

A `for` loop is used to iterate over elements of a sequence (such as a list, string, or array) one by one until the sequence is exhausted.

### 2.1 Exercise 1: Basic for Loop over a List
* **Objective:** Loop through a list of room areas and print each element separately.


```python
# areas list
areas = [11.25, 18.0, 20.0, 10.75, 9.50]

# Code the for loop
for area in areas:
    print(area)
```

### 2.2 Exercise 2 & 3: Using `enumerate()` for Indexes and Values
* **Objective:** Use `enumerate()` to access both the element's index and its value simultaneously, and format the output.
  

```python
# Change for loop to use enumerate() and start counting from 1
for index, area in enumerate(areas):
    print("room " + str(index + 1) + ": " + str(area))
```

### 2.3 Exercise 4: Looping over a List of Lists
* **Objective:** Iterate through a nested list structure, extracting specific sub-elements (room name and area) to build a custom string.


```python
# house list of lists
house = [["hallway", 11.25], 
         ["kitchen", 18.0], 
         ["living room", 20.0], 
         ["bedroom", 10.75], 
         ["bathroom", 9.50]]

# Build a for loop from scratch
for room in house:
    print("the " + str(room[0]) + " is " + str(room[1]) + " sqm")
```

## 3. Loop Data Structures (Part 1)

In this section, we explore how to iterate over more complex data structures, specifically **Dictionaries** and **NumPy Arrays**, which require specialized methods and functions to access their underlying data.

### 3.1 Exercise 1: Looping over a Dictionary
* **Objective:** Use the `.items()` method to loop through a dictionary and extract both the keys (countries) and values (capitals) simultaneously.


```python
# Definition of dictionary
europe = {'spain':'madrid', 'france':'paris', 'germany':'berlin', 
          'norway':'oslo', 'italy':'rome', 'poland':'warsaw', 'austria':'vienna' }
          
# Iterate over europe
for key, value in europe.items():
    print("the capital of " + str(key) + " is " + str(value))
```



### 3.2 Exercise 2: Looping over NumPy Arrays
* **Objective:** Iterate through a 1D NumPy array using a standard `for` loop, and use `np.nditer()` to loop through every individual element of a 2D array.

```python
import numpy as np

# Creating sample data to match the exercise environment
np_height = np.array([74, 74, 72, 72, 73])
np_baseball = np.array([[74, 180], [74, 215], [72, 210], [72, 210], [73, 188]])

# For loop over np_height (1D Array)
for x in np_height:
    print(str(x) + " inches")
    
print("-" * 20) # Separator Line

# For loop over np_baseball (2D Array) using np.nditer()
for x in np.nditer(np_baseball):
    print(x)
```

## 4. Loop Data Structures (Part 2) - Pandas DataFrames

When iterating over a Pandas DataFrame, using a standard `for` loop only returns the column names. To iterate through the actual data row by row, specialized Pandas methods like `.iterrows()` or the optimized `.apply()` function must be used.

### 4.1 Exercise 1 & 2: Iterating Row by Row using `.iterrows()`
* **Objective:** Use `.iterrows()` to loop through the `cars` DataFrame, accessing the row labels and specific column values (such as `cars_per_cap`) to print formatted statistics.


```python
import pandas as pd
# Load the dataset
cars = pd.read_csv('../datasets/cars.csv', index_col=0)
cars
```

```python
# Iterate over rows of cars and print formatted string
for lab, row in cars.iterrows():
    print(str(lab) + ": " + str(row["cars_per_cap"]))
```



### 4.2 Exercise 3 & 4: Adding New Columns (`.iterrows()` vs `.apply()`)
* **Objective:** Compare adding a new mutated column (`COUNTRY`) using a slow sequential `for` loop versus using the highly optimized and vectorized `.apply()` function.


```python
# 1. Using the advanced and vectorized .apply() method (Best Practice)
cars["COUNTRY"] = cars["country"].apply(str.upper)

# Display the updated DataFrame
print(cars)
```
