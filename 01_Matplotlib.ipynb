## Lesson 1: Introduction to Line Plots

This notebook is part of my **Data Analyst with Python** learning path. In this lesson, I explore the fundamentals of data visualization using the `matplotlib` library, focusing specifically on creating and customizing line plots to track global population growth trends.

---

### Objectives:
* Import `matplotlib.pyplot` package.
* Create a basic line plot with specified X and Y axes.
* Display plots using the `plt.show()` function.
* Analyze historical world population data.




A **Line Plot** is a type of chart that displays information as a series of data points connected by straight line segments. It is best used to track changes or trends over time (chronological data).


```python
# Step 1: Import the matplotlib library
import matplotlib.pyplot as plt

# Step 2: Define the data (Example data for testing on your device)
year = [1950, 1970, 1990, 2010, 2100]
pop = [2.5, 3.6, 5.2, 6.9, 10.8]

# Step 3: Create a line plot year = x , pop = y
plt.plot(year, pop)

# Step 4: Display the plot
plt.show()
```

## Lesson 1.2: Scatter Plots

In this section, I explore **Scatter Plots** using `plt.scatter()`. 
Unlike line plots, scatter plots display data as individual points. This is ideal for finding relationships or correlations between two continuous variables without implying a chronological order.

### Key Features Addressed:
* **Function**: `plt.scatter(x, y)`
* **Logarithmic Scaling**: Used `plt.xscale('log')` to de-clutter data and handle heavily skewed values (like population or GDP per capita).


```python
# Step 1: Define variables for the scatter plot
gdp_cap = [1000, 2000, 5000, 10000, 20000, 30000, 40000, 50000, 60000]
life_exp = [43.8, 58.1, 60.1, 65.2, 72.3, 74.7, 75.3, 76.4, 78.5]

# Step 2: Build the scatter plot
plt.scatter(gdp_cap, life_exp)

# Step 3: Change X-axis to logarithmic scale
plt.xscale('log')

# Step 4: Display the plot
plt.show()
```

## Lesson 1.3: Histograms & Data Distributions

In this section, I explore **Histograms** using `plt.hist()`. Unlike line and scatter plots, a histogram visualizes the distribution of a single dataset by dividing data into continuous intervals called **bins**. 

I practiced altering the `bins` parameter (e.g., 5, 10, 15, and 20) to see how data granularity changes. Additionally, I learned to use `plt.clf()` to clear the current figure canvas, allowing me to compare the global life expectancy distribution of 1950 against 2007 across separate, clean plots.


```python
# Step 1: Define datasets (Example data for 1950 and 2007)
life_exp = [43.8, 76.4, 72.3, 65.2, 58.1, 74.7, 60.1, 75.3, 78.5, 46.9]
life_exp1950 = [33.2, 54.1, 48.3, 40.2, 38.5, 62.1, 45.4, 55.2, 60.3, 35.1]

# Step 2: Plot the 2007 Histogram (with 10 bins)
plt.hist(life_exp, bins=10, edgecolor='black')
plt.show()

# Step 3: Clear the figure canvas to avoid overlaying plots
plt.clf()

# Step 4: Plot the 1950 Histogram on a fresh canvas
plt.hist(life_exp1950, bins=10, edgecolor='black', color='orange')
plt.show()
```

## Lesson 1.4: Advanced Customizations (Bubble Charts, Labels, and Grids)

In this section, I advanced the `plt.scatter()` plot into a comprehensive **Bubble Chart** by customizing several plot properties to inject more data layers into a single view:
* **Sizes (`s=np_pop`)**: Used a `numpy` array to tie bubble sizes directly to country population data, making highly populated nations visually stand out.
* **Colors (`c=col`) & Opacity (`alpha=0.8`)**: Distinctly mapped colors based on geographic continents and introduced transparency to unmask overlapping data points.
* **Annotations (`plt.text`)**: Positioned direct text labels on the grid coordinates for specific landmarks like *India* and *China*.
* **Grids (`plt.grid(True)`)**: Added background reference lines to significantly improve data readability. 
m

```python
import numpy as np

# 1. Dummy Data for testing on your device
gdp_cap = [974, 5937, 6223, 4797, 12779, 35318, 1055, 33203, 42951, 10000]
life_exp = [43.8, 76.4, 72.3, 65.2, 58.1, 74.7, 60.1, 75.3, 78.5, 71.0]
pop = [31.8, 12.8, 1148.0, 1318.0, 28.7, 30.3, 21.4, 304.0, 3.8, 50.0]

# Pre-defined colors for each point based on continents (Example)
col = ['blue', 'green', 'red', 'red', 'blue', 'green', 'blue', 'orange', 'green', 'orange']

# 2. Convert pop to numpy array and multiply size for clear visibility
np_pop = np.array(pop) * 0.8  # Adjusted scaling factor for small dummy list

# 3. Build Bubble chart with custom colors and transparency
plt.scatter(gdp_cap, life_exp, s=np_pop, c=col, alpha=0.8)

# 4. Customizations
plt.xscale('log') 
plt.xlabel('GDP per Capita [in USD]')
plt.ylabel('Life Expectancy [in years]')
plt.title('World Development in 2007 (Customized)')
plt.xticks([1000, 10000, 100000], ['1k', '10k', '100k'])

# 5. Add Text Annotations (Positioned based on dummy data coordinates)
plt.text(1318, 65, 'India')
plt.text(1148, 72, 'China')

# 6. Turn on grid lines
plt.grid(True)

# 7. Display final polished plot
plt.show()
```
